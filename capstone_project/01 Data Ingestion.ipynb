```python
%sql

CREATE SCHEMA IF NOT EXISTS bronze;
```

###SQL Data Ingestion

```python
customers_df = spark.read.csv(path='/Volumes/shopsphere/ecomm_schema/capstone_data/sql_server/initial/customers.csv', header=True, inferSchema=True)

order_items_df = spark.read.csv(path='/Volumes/shopsphere/ecomm_schema/capstone_data/sql_server/initial/order_items.csv', header=True, inferSchema=True)

orders_df = spark.read.csv(path='/Volumes/shopsphere/ecomm_schema/capstone_data/sql_server/initial/orders.csv', header=True, inferSchema=True)

payments_df = spark.read.csv(path='/Volumes/shopsphere/ecomm_schema/capstone_data/sql_server/initial/payments.csv', header=True, inferSchema=True)
```

```python
customers_df.write.saveAsTable('bronze.customers',mode='OVERWRITE')
order_items_df.write.saveAsTable('bronze.order_items',mode='OVERWRITE')
orders_df.write.saveAsTable('bronze.orders',mode='OVERWRITE')
payments_df.write.saveAsTable('bronze.payments',mode='OVERWRITE')
```

###ADLS Data Ingestion

```python
products_df = spark.read.csv(path='/Volumes/shopsphere/ecomm_schema/capstone_data/adls/initial/products/', header=True, inferSchema=True)

inventory_df = spark.read.csv(path='/Volumes/shopsphere/ecomm_schema/capstone_data/adls/initial/inventory/', header=True, inferSchema=True)

clickstream_df = spark.read.json(path='/Volumes/shopsphere/ecomm_schema/capstone_data/adls/initial/clickstream/')
```

```python
products_df.write.saveAsTable('bronze.products',mode='OVERWRITE')
inventory_df.write.saveAsTable('bronze.inventory',mode='OVERWRITE')
clickstream_df.write.saveAsTable('bronze.clickstream',mode='OVERWRITE')
```
